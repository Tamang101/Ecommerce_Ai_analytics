```python
# Import Dependencies

import pandas as pd
```

1. Data Cleaning

```python
# Opening Dataset

df = pd.read_excel('C:/Users/asus/Downloads/Bank_dataset/online_retail_II.xlsx')
```

```python
df.head()
```

```python
# info of data set 

df.info()
```

```python
# checking for the  missing values 

print(df.isnull().sum())
```

```python
# Hnadling missing Customer ID

# In terms of RFM We need to drop our customer Id with missing value because it will unable t0 grouping imposible 

df = df.dropna(subset = ['Customer ID'])
```

```python
df.isnull().sum()
```

```python
print(df['Description'].isnull().sum())
```

```python
print(df['Description'].isnull().sum())
```

```python
# Remove duplicates 

df.duplicated().sum()
```

```python
Duplicates = df.apply(lambda x: x.duplicated().sum())
```

```python
Duplicates
```

```python
df = df.drop_duplicates()
```

```python
# Recheck it 
```

```python
df.duplicated().sum()
```

```python
df['InvoiceDate'].head()
```

```python
# still not changed so we will make a copy of our Data Frame 
df = df.copy()  # now df is a full independent DataFrame

df['Year']  = pd.to_datetime(df['InvoiceDate']).dt.year
df['Month'] = pd.to_datetime(df['InvoiceDate']).dt.month
df['Day']   = pd.to_datetime(df['InvoiceDate']).dt.day
df['Hour']  = pd.to_datetime(df['InvoiceDate']).dt.hour
```

```python
df['InvoiceDate'].head()
```

```python
# For the monthly aggregation we need a new column called  ' InvoiceMonth '
df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')
```

```python
df['InvoiceMonth'].head()
```

```python
# Remove cancelled/refunded transactions (negative Quantity)

original_count = len(df)
df = df[df['Quantity']> 0].copy()
removed_count = original_count - len(df)
print(f"Removed {removed_count} cancelled/refunded transactions")
```

```python
df.head()
```

```python
# Saving Cleanded Data

file_path = r'C:\Users\asus\Downloads\Bank_dataset\cleaned_data.xlsx'
df.to_excel(file_path, index=False, engine='openpyxl')

```
